```python
import pandas as pd
import os

dataset_path = r"C:\Users\admin\Lab8_Data_Visualization_Celo\spotify_dataset\spotify_top_1000_tracks.csv"

df = pd.read_csv(dataset_path, encoding="utf-8")
```

```python
# [OUTPUT] A table showing track name, artist, and duration.

# Sort the dataset by duration_min in descending order.     Select the top 10 longest tracks.
top_10 = df.sort_values(by='duration_min', ascending=False).head(10)

# Changed the cases
table = top_10[['track_name', 'artist', 'duration_min']].rename(columns={
    'track_name': 'Track Name',
    'artist': 'Artist',
    'duration_min': 'Duration (min)'
})

print("=" * 31 + " Top 10 Longest Tracks " + "=" * 31)
print("\n" + table.to_string(index=False))
```

```python
# [OUTPUT] A horizontal bar chart visualizing track duration.

import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

# Create a bar chart with track names on the y-axis and durations on the x-axis, colored using the student-specific color.
plt.barh(top_10['track_name'], top_10['duration_min'], color=color1)
plt.title(f"Top 10 Longest Songs in Spotify Top 1000", fontsize=14, fontname='Times New Roman')
plt.xlabel("Duration (minutes)", fontsize=12, fontname='Times New Roman')
plt.ylabel("Track Name", fontsize=12, fontname='Times New Roman')
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.show()
```
